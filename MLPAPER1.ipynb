```python
class_counts = df['class'].value_counts()
colors = ['skyblue', 'orange']
plt.figure(figsize=(7, 7))
plt.pie(class_counts, labels=class_counts.index, autopct='%1.1f%%', startangle=0,colors=colors)
hfont = {'fontname':'serif', 'weight': 'bold'}
plt.title('Label', size=15,**hfont)
plt.legend()

plt.show()
```

```python
from sklearn.preprocessing import LabelEncoder
label_encoder = LabelEncoder()
df['Category'] = label_encoder.fit_transform(df['Category'])
df['Package'] = label_encoder.fit_transform(df['Package'])
```

```python
c=['Package','Category','class']
```

```python
sdf = df.drop(c,axis=1)
```

```python
from sklearn.decomposition import PCA
n_components = 30
pca = PCA(n_components=n_components)
```

```python
pca.fit(sdf)
s_transformed = pca.transform(sdf)


column_names = ['Vpca_column_{}'.format(i+1) for i in range(30)]
s_transformed_df=pd.DataFrame(s_transformed, columns=column_names)
```

```python
s_transformed_df['Package']=df['Package']
s_transformed_df['Category']=df['Category']
s_transformed_df['class']=df['class']
```

```python
X=df.drop(['class'],axis=1).values
y=df['class'].values
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.33, random_state=42)
```

```python
from sklearn.ensemble import RandomForestClassifier, ExtraTreesClassifier
```

```python
rf = RandomForestClassifier(random_state=0,n_estimators=500,min_samples_split=100)
rf.fit(X_train, y_train)
```

```python
rf_pred = rf.predict(X_test)
rf_pred2 = rf.predict(X_train)
```

```python
from sklearn.metrics import accuracy_score, classification_report, confusion_matrix, ConfusionMatrixDisplay
```

```python
print("train_accuracy:")
print(accuracy_score(rf_pred2,y_train)*100)
print('----------------------------------')
print("test_accuracy:")
print(accuracy_score(rf_pred,y_test)*100)
print('----------------------------------')
print(classification_report(y_test, rf_pred))
```

```python
eet =ExtraTreesClassifier()
eet.fit(X_train, y_train)
```

```python

eet_pred = eet.predict(X_test)
eet_pred2 = eet.predict(X_train)
```

```python
print("train_accuracy:")
print(accuracy_score(eet_pred2,y_train)*100)
print('----------------------------------')
print("test_accuracy:")
print(accuracy_score(eet_pred,y_test)*100)
print('----------------------------------')
print(classification_report(y_test, eet_pred))
print('----------------------------------')
```

```python
from sklearn.tree import DecisionTreeClassifier, ExtraTreeClassifier
```

```python
dt_pred = dt.predict(X_test)
dt_pred2 = dt.predict(X_train)
```
